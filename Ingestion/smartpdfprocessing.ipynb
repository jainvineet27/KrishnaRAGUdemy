### RecursiveCharacterTextSplitter
- Tumne arguments positionally diye
- RecursiveCharacterTextSplitter keyword arguments expect karta hai
- Isliye Python confuse ho gaya → error

```python
class P:
    def __init__(self, chunk_size=1000,  chunk_overlap=100):
        self.chunk_size =chunk_size 
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(chunk_size=self.chunk_size, 
                                                            chunk_overlap=self.chunk_overlap , separators=[" "])
p =P()
p.chunk_size
p.chunk_overlap
p.text_splitter._separators
```
