```python
from pydantic import BaseModel , Field
from typing import List, Optional
import os
```

```python
class Person(BaseModel):
    name:str 
    age:int 
    city:str


p =Person(name='vin', age=12, city='Hyd')  
print(p)
p2 =Person(name='vineet', age=22, city='Hyd')
```

### Pydantic with Optional

```python
class Person(BaseModel):
    name:str 
    age:int 
    city:str
    is_active : Optional[bool] = False  # Here we have not applied any default value so error 

p =Person(name='vin', age=12, city='Hyd')  
print(p)
p2 =Person(name='vineet', age=22, city='Hyd')
```

#### Enforcing strict and other constraints

```python
class Users(BaseModel):
    names:List[str]=Field(strict=True, min_length=1 , max_length =33)
    ages:List[int]=Field(min_length=1 , max_length =33)

users = Users(names=['vin', 'vineet'], ages=[12, 22])
print(users)
try:
    users2 = Users(names=['vin', 'vineet', 11], ages=[12, 22, 33])
    print(users2)
except Exception as e:
    print(e, "https://pydantic.dev/docs/validation/latest/errors/validation_errors/#string_type")

```

```python
import requests 
import  os 
```

```python
NASA_API_KEY =os.getenv("NASA_API_KEY")
url =f"https://api.nasa.gov/planetary/apod?api_key={NASA_API_KEY}"
response = requests.get(url)

clear_output = response.json()
#clear_output...
for k,v in clear_output.items():
    print(k)
```

```python
from pydantic import ValidationError
from datetime import date
class NasaDataValidation(BaseModel):
    date:date
    explanation:str
    hdurl:str
    media_type:str
    service_version:str
    title:str
    url:str

```

```python
try:    
    nasadata = NasaDataValidation(**clear_output)
    print('Validation successful!')
    print(nasadata)

except ValidationError as e :
    print(e)
```
