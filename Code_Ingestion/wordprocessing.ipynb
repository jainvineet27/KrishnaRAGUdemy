```python
from langchain_community.document_loaders import Docx2txtLoader , UnstructuredWordDocumentLoader
```

```python
import os

# current working directory
cwd = os.getcwd()

# go exactly one folder back (without changing CWD)
parent = os.path.abspath(os.path.join(cwd, ".."))
print(f"Current working directory: {cwd}")
print(f"Parent directory: {parent}")
```

```python

file_name = "LOR - by Ms. Sonia Jain.docx"
file_path = os.path.join(parent, "otherfile", file_name)
print(f"File path: {file_path}")
```

```python
docx = Docx2txtLoader(file_path)
docs =docx.load()
print(len(docs))
print(docs[0].page_content[:40])
```

```python
unstructured = UnstructuredWordDocumentLoader(file_path=file_path, mode="elements")
docs =unstructured.load()
print(docs[0].page_content[0:40])
print(len(docs))

for i , doc in enumerate(docs):
    print(f"\n Element {i+1}")
    print(f"Type {doc.metadata.get("category",'')} ")
    print(f"{doc.page_content[:40]}")
```
