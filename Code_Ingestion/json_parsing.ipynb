### f.write()   ye sirft string smajahata hai  python dictionary or other object nahi samjahta 

```python
API_KEY = os.getenv("NASA_API_KEY")
url = "https://api.nasa.gov/planetary/apod?api_key"

try: 
    response = requests.get(f'{url}={API_KEY}')

    output  =response.json()  #this interanll  loads(response.text) the json data into a python dictionary
    print(output)
    for k,v in output.items():
        print(f'Key==> {k} ,  Value ==> {v}')
    file_name = "api.json"
    folder_name= "otherfile/json"
    base_path = os.path.abspath(os.path.join(os.getcwd(), '..'))
    file_path  = os.path.join(base_path,folder_name , file_name)
    if not os.path.exists(file_path):
        with open(file_path,'w') as f:
            json.dump(response.text,f)     # maine ab json string ko file me dump kar diya hai, 
                                    #kyuki response.text me json string hoti hai, aur json.dump usko file me likhne ke liye use hota hai
            print('File data is printed successfuly..')
    else:
        print('file exists')


except Exception as e : 
    print(f'Error is {e}')
```

```python
from langchain_community.document_loaders import JSONLoader
```

```python
jsonobject = JSONLoader(file_path=file_path, jq_schema=".explanation")
doucments=jsonobject.load()
print(len(doucments))
print(doucments[0].metadata)
print(doucments[0].page_content)


```
