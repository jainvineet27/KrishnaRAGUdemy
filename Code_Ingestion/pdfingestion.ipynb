```python
from langchain_community.document_loaders import (
    PyPDFLoader,
    PyMuPDFLoader,
    UnstructuredPDFLoader,
)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
# current working directory
cwd = os.getcwd()

# go exactly one folder back (without changing CWD)
parent = os.path.abspath(os.path.join(cwd, ".."))
print(f"Current working directory: {cwd}")
print(f"Parent directory: {parent}")
```

```python

file_name = "attention.pdf"
file_path = os.path.join(parent, "otherfile", file_name)
print(f"File path: {file_path}")
```

```python
try:        
    print(f"File path: {file_path}")
    if os.path.exists(file_path):
        print('FileExistsError',end=' ' )
        pyPDFLoader = PyPDFLoader(file_path)
        docs = pyPDFLoader.load()
        #i revied list of documents
        print(len(docs))
        for doc in docs:
            print(type(doc))
            print(doc.page_content[:100])
        # for k,v in doc.metadata.items():
        #     print(f"{k}: {v}")
    else:
        print(f"File {file_path} does not exist.")

except Exception as e:
    print(f'Error has occured {e}')
```

```python
try:
    pymuPDFLoader = PyMuPDFLoader(file_path)
    documents =  pymuPDFLoader.load()

    print(f'Number of documents: {len(documents)}')
    print(f'Type of first document: {type(documents[0])}')
    #rint(f'Content of first document: {documents[0].page_content[:100]}')
    print(f'Metadata of first document: {documents[0].metadata}')
except Exception as e:
    print(f'Error has occured: {e}')
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100,
    separators=["\n\n", "\n", ".", " ", ""]
)

chunks = splitter.split_documents(docs) 

print(len(chunks))
print(chunks[0].page_content[:300])

for i in range(2):
    print(chunks[i], '>>>>>>>>>>' , docs[i])
```
